This notebook shows step-by-step calculations using SymPy for key equations.

## Section: Standardisasi Data

### Equation (2.1) (Label: `eq:standardization`)

\begin{equation}
z_{ij} = \frac{x_{ij} - \bar{x}_j}{s_j}, \quad i = 1, 2, \ldots, n, \quad j = 1, 2, \ldots, p
\end{equation}

```python
# Step-by-step: Z-Score Standardization
from IPython.display import display

# 1. Define symbolic variables
x_ij, x_bar, s_j, z_ij = sp.symbols('x_{ij} \bar{x}_j s_j z_{ij}')

# 2. State the formula
formula = sp.Eq(z_ij, (x_ij - x_bar) / s_j)
print("Step 1: Defined the standardization formula")
display(formula)

# 3. Example values
subs_dict = {x_ij: 150, x_bar: 100, s_j: 25}
print(f"Step 2: Input values: x={subs_dict[x_ij]}, mean={subs_dict[x_bar]}, std={subs_dict[s_j]}")

# 4. Substitution step
step_substituted = sp.Eq(z_ij, (subs_dict[x_ij] - subs_dict[x_bar]) / subs_dict[s_j], evaluate=False)
display(step_substituted)

# 5. Final Calculation
result = (subs_dict[x_ij] - subs_dict[x_bar]) / subs_dict[s_j]
print(f"Step 3: Resulting Z-score: {result}")
```

## Section: Ukuran Pemusatan

### Equation (2.2) (Label: `eq:mean`)

\begin{equation}
\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i
\end{equation}

```python
# Step-by-step: Mean Calculation
n = sp.symbols('n')
x = sp.IndexedBase('x')
i = sp.symbols('i', cls=sp.Idx)
mean = sp.symbols('\bar{x}')

# 1. General Formula
gen_formula = sp.Eq(mean, (1/n) * sp.Sum(x[i], (i, 1, n)))
print("Step 1: Formula definition")
display(gen_formula)

# 2. concrete example
n_val = 5
values = [10, 20, 30, 40, 50]
print(f"Step 2: Data set: {values}, n={n_val}")

# 3. Summation step
sum_val = sum(values)
print(f"Step 3: Calculate Sum = {sum_val}")

# 4. Final step
final_mean = sum_val / n_val
print(f"Step 4: Mean = {sum_val} / {n_val} = {final_mean}")
```

## Section: Tahap 1: Intra-struktur

### Equation (2.25) (Label: `eq:intra`)

\begin{equation}
\mathbf{W}_k = \mathbf{Z}_k \mathbf{Z}_k^T
\end{equation}

```python
# Step-by-step: Intra-structure (Scalar Product Matrix)
# W_k = Z_k * Z_k^T

# 1. Define Symbols
Z_k = sp.MatrixSymbol('Z_k', 3, 2) # Example 3 objects, 2 variables
W_k = sp.MatrixSymbol('W_k', 3, 3)

formula = sp.Eq(W_k, Z_k * Z_k.T)
print("Step 1: Relationship between standardized data Z and Operator W")
display(formula)

# 2. Example Data
Z_val = sp.Matrix([[1, 2], [3, 4], [5, 6]])
print("Step 2: Example Z values (3 objects, 2 variables):")
display(Z_val)

# 3. Transpose
print("Step 3: Calculating Transpose Z^T:")
display(Z_val.T)

# 4. Multiplication
W_val = Z_val * Z_val.T
print("Step 4: Resulting W matrix (Similarity between objects):")
display(W_val)
```

## Section: Tahap 2: Inter-struktur

### Equation (2.26) (Label: `eq:interstructure_H`)

\begin{equation}
H_{kk'} = \mathrm{tr}(\mathbf{W}_k \mathbf{W}_{k'}), \qquad k,k' = 1,\ldots,K.
\end{equation}

```python
# Step-by-step: Inter-structure (Hilbert-Schmidt)
# H_kk' = tr(W_k * W_k')

W1 = sp.Matrix([[1, 0], [0, 1]])
W2 = sp.Matrix([[2, 1], [1, 2]])

print("Step 1: Matrix W for situation 1:")
display(W1)
print("Step 2: Matrix W for situation 2:")
display(W2)

# 3. Product
prod = W1 * W2
print("Step 3: Matrix Product W1 * W2:")
display(prod)

# 4. Trace
trace_val = prod.trace()
print(f"Step 4: Matrix Trace (Sum of Diagonal) = {trace_val}")
```

### Equation (2.27) (Label: `eq:rv`)

\begin{equation}
RV(\mathbf{W}_k, \mathbf{W}_{k'}) = \frac{\mathrm{tr}(\mathbf{W}_k \mathbf{W}_{k'})}{\sqrt{\mathrm{tr}(\mathbf{W}_k^2) \, \mathrm{tr}(\mathbf{W}_{k'}^2)}}
    = \frac{H_{kk'}}{\sqrt{H_{kk}\,H_{k'k'}}}.
\end{equation}

```python
# Step-by-step: RV Coefficient
# RV = tr(W1*W2) / sqrt(tr(W1^2) * tr(W2^2))

def hs_product(A, B): return (A * B).trace()

W1 = sp.Matrix([[1, 0], [0, 1]])
W2 = sp.Matrix([[0.8, 0.1], [0.1, 0.8]])

num = hs_product(W1, W2)
den1 = hs_product(W1, W1)
den2 = hs_product(W2, W2)

print(f"Step 1: Scalar Product (Numerator) = {num}")
print(f"Step 2: Norm situation 1 = {sp.sqrt(den1)}")
print(f"Step 3: Norm situation 2 = {sp.sqrt(den2)}")

rv = num / sp.sqrt(den1 * den2)
print(f"Step 4: RV Coefficient = {float(rv)}")
```
